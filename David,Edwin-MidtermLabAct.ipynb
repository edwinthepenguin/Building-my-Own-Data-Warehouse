**Create Your Own SQL Questions**

Question: How many individual transactions were conducted by each customer type?

````python
%sql
SELECT 
    c.customer_type, 
    COUNT(f.transaction_id) AS total_transactions 
FROM fact_sales f 
JOIN dim_customer c ON f.customer_id = c.customer_id 
GROUP BY c.customer_type;
````

Interpretation: Retailers and wholesale buyers form a significant portion of our recorded transaction frequency, showing steady day-to-day order counts compared to institutional or bar accounts.

Recommendation: Implement a loyalty discount program specifically for high-frequency customer types like Retailers and Wholesalers to encourage bulk reordering.

Question: What is the total gross revenue generated by each product?

````python
%sql
SELECT 
    p.product_name, 
    SUM(f.total_sales) AS gross_revenue 
FROM fact_sales f 
JOIN dim_product p ON f.product_id = p.product_id 
GROUP BY p.product_name 
ORDER BY gross_revenue DESC;
````

Interpretation: Red Horse Extra Strong Beer generates the vast majority of total business revenue, significantly outperforming soft drinks and other spirits due to high-volume case sales.

Recommendation: Ensure safety stock levels for Red Horse Extra Strong Beer are maintained well above average to prevent stockouts during peak weekends and holidays.

Question: What is the average number of boxes purchased per transaction across all sales?

````python
%sql
SELECT 
    AVG(quantity_boxes) AS avg_boxes_per_transaction 
FROM fact_sales;
````

Interpretation: On average, customers purchase roughly 10 boxes per single transaction order, reflecting a healthy mid-tier wholesale and retail distribution volume.

Recommendation: Set a minimum order threshold of 15 boxes for free delivery to incentivize customers to buy slightly above the current transaction average.

Question: What is the total box volume sold for each product category?

````python
%sql
SELECT 
    p.category, 
    SUM(f.quantity_boxes) AS total_boxes_sold 
FROM fact_sales f 
JOIN dim_product p ON f.product_id = p.product_id 
GROUP BY p.category;
````

Interpretation: The Beer category dominates total box movement by a wide margin, proving that alcoholic beverages drive the primary bulk volume of our distribution business.

Recommendation: Bundle slower-moving soft drinks with high-demand beer cases in promotional packages to cross-sell inventory.

Question: Which five individual transactions yielded the highest total sales amount?

````python
%sql
SELECT 
    transaction_id, 
    total_sales 
FROM fact_sales 
ORDER BY total_sales DESC 
LIMIT 5;
````

Interpretation: The 4 top-grossing transactions all exceed 20,000 PHP, driven almost entirely by massive bulk purchases of high-priced liquor and beer cases by commercial buyers.

Recommendation: Assign dedicated account managers to handle these top-tier commercial buyers to ensure long-term retention.

Question: Which customers bought which products, and on what dates did those transactions occur?

````python
%sql
SELECT 
    c.customer_name, 
    p.product_name, 
    dt.full_date, 
    f.quantity_boxes, 
    f.total_sales
FROM fact_sales f
JOIN dim_customer c ON f.customer_id = c.customer_id
JOIN dim_product p ON f.product_id = p.product_id
JOIN dim_date dt ON f.date_id = dt.date_id
LIMIT 5;
````

Interpretation: This query successfully maps transactional facts across our three reference dimensions, revealing exact purchasing habits per customer over time.

Recommendation: Use this multi-table join view to build customer purchase history profiles for targeted marketing and re-order reminders.

Question: What was the total revenue generated exclusively during the month of January 2026?

````python
%sql
SELECT 
    dt.month_name, 
    SUM(f.total_sales) AS january_revenue
FROM fact_sales f
JOIN dim_date dt ON f.date_id = dt.date_id
WHERE dt.month_name = 'January' AND dt.year = 2026
GROUP BY dt.month_name;
````

Interpretation: January 2026 established a steady baseline of initial sales activity across our active customer network.

Recommendation: Compare January's performance against subsequent months to evaluate seasonal growth trends and peak demand periods.

Question: How can products be classified into performance tiers (High-Performing, Medium-Performing, Low-Performing) based on their total boxes sold?

````python
%sql
SELECT 
    p.product_name,
    SUM(f.quantity_boxes) AS total_boxes,
    CASE 
        WHEN SUM(f.quantity_boxes) >= 30 THEN 'High-Performing'
        WHEN SUM(f.quantity_boxes) BETWEEN 10 AND 29 THEN 'Medium-Performing'
        ELSE 'Low-Performing'
    END AS performance_tier
FROM fact_sales f
JOIN dim_product p ON f.product_id = p.product_id
GROUP BY p.product_name
ORDER BY total_boxes DESC;
````

Interpretation: The categorization clearly separates our blockbuster inventory items from slow movers, identifying which beverages anchor our revenue.

Recommendation: Phase out or reduce purchase orders for persistent "Low-Performing" stock items to free up warehouse storage space for high-demand products.

Part H — Advanced SQL Requirement

````python
%sql
WITH monthly_summary AS (
    SELECT 
        dt.month_name,
        dt.month_number,
        SUM(f.total_sales) AS current_monthly_sales
    FROM fact_sales f
    JOIN dim_date dt ON f.date_id = dt.date_id
    GROUP BY dt.month_name, dt.month_number
)
SELECT 
    month_name,
    current_monthly_sales,
    LAG(current_monthly_sales) OVER (ORDER BY month_number) AS previous_month_sales,
    (current_monthly_sales - LAG(current_monthly_sales) OVER (ORDER BY month_number)) AS revenue_difference
FROM monthly_summary
ORDER BY month_number;
````

````python
%python
from pyspark.sql import functions as F

# Load the fact table into a DataFrame
sales_df = spark.table("fact_sales")

# Filter for bulk orders where quantity is 15 boxes or more
bulk_orders_df = sales_df.filter(F.col("quantity_boxes") >= 15)

# Display the filtered DataFrame
display(bulk_orders_df)
````

````python
%python
from pyspark.sql import functions as F

sales_df = spark.table("fact_sales")

# Group by customer_id and aggregate performance metrics
customer_summary = (
    sales_df.groupBy("customer_id")
    .agg(
        F.sum("quantity_boxes").alias("total_boxes_bought"),
        F.sum("total_sales").alias("total_spent"),
        F.count("transaction_id").alias("order_count")
    )
    .orderBy(F.desc("total_spent"))
)

display(customer_summary)
````

````python
%python
from pyspark.sql import functions as F

sales_df = spark.table("fact_sales")
product_df = spark.table("dim_product")

# Join sales with product dimension and aggregate by category
category_analysis = (
    sales_df.join(product_df, sales_df.product_id == product_df.product_id, "inner")
    .groupBy("category")
    .agg(
        F.sum("total_sales").alias("category_revenue"),
        F.sum("quantity_boxes").alias("category_boxes_sold")
    )
    .orderBy(F.desc("category_revenue"))
)

display(category_analysis)
````

````python
%python
from pyspark.sql import functions as F

sales_df = spark.table("fact_sales")
product_df = spark.table("dim_product")

# Join, group, and aggregate using PySpark DataFrames
category_comparison = (
    sales_df.join(product_df, sales_df.product_id == product_df.product_id, "inner")
    .groupBy("category")
    .agg(
        F.sum("total_sales").alias("total_revenue"),
        F.sum("quantity_boxes").alias("total_boxes_sold")
    )
    .orderBy(F.desc("total_revenue"))
)

display(category_comparison)
````

What error could cause the two results to be different?
